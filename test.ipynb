### Pengantar
Dalam dunia nyata, data sering disimpan secara permanen dalam bentuk file, bukan hanya di memori program.[cite: 1]
Python menyediakan fitur File Handling untuk berinteraksi dengan file membaca, menulis, dan memperbarui data yang tersimpan di disk.[cite: 1]
Setiap file memiliki dua bagian utama:
* Path (lokasi) - tempat file berada.[cite: 1]
* Isi (content) - data yang tersimpan di dalam file.[cite: 1]

File dapat berupa teks (.txt, .csv) atau biner (.jpg, .exe, .pdf).[cite: 1]
Python mengakses file menggunakan fungsi open ().[cite: 1]

### A. Fungsi Dasar File Handling
**1. Membuka File**[cite: 1]
Sintaks dasar: `file = open("nama file.txt", "mode")`[cite: 1]

**2. Menutup File**[cite: 2]
Setiap file yang dibuka harus ditutup: `file.close()`[cite: 2]
Namun lebih aman menggunakan `with open()` agar file otomatis ditutup:[cite: 2]
```python
with open ("data.txt", "r") as file:
    isi = file.read()
```[cite: 2]

### B. Mode Akses File
Sebelum bekerja dengan file, kita perlu menentukan tujuan pembukaan file tersebut, apakah untuk membaca isinya, menulis data baru, atau menambah data ke file yang sudah ada.[cite: 2]
Tujuan ini disebut mode akses file (file access mode).[cite: 2]
Mode ini memberi tahu Python bagaimana cara memperlakukan file misalnya, hanya untuk dibaca (read-only), menulis dari awal (overwrite), atau menambah di akhir file (append).[cite: 2]
Dengan menggunakan mode yang tepat, kita dapat mencegah kehilangan data atau kesalahan saat memproses file.[cite: 2]

| Mode | Fungsi | Deskripsi |
|---|---|---|
| 'r' | Read | Membaca file (error jika tidak ada) |[cite: 2]
| 'w' | Write | Menulis file baru, menimpa jika sudah ada |[cite: 2]
| 'a' | Append | Menambahkan data di akhir file |[cite: 2]
| 'r+' | Read & Write | Membaca dan menulis |[cite: 2]
| 'b' | Binary | Mode biner, digunakan untuk file non-teks |[cite: 2]

### C. Metode Umum File Object
Setelah sebuah file dibuka menggunakan fungsi open(), Python akan menghasilkan objek file yaitu representasi file tersebut di dalam memori.[cite: 2]
Objek file ini memiliki berbagai metode (fungsi bawaan) yang digunakan untuk melakukan operasi terhadap isi file, seperti membaca seluruh isi file, membaca baris tertentu, menulis teks, atau menutup file setelah selesai digunakan.[cite: 2]
Dengan memahami metode-metode ini, kita dapat mengontrol bagaimana data dibaca atau ditulis ke dalam file secara efisien dan aman..[cite: 2]

| Metode | Deskripsi |
|---|---|
| read() | Membaca seluruh isi file |[cite: 2]
| readline() | Membaca satu baris dari file |[cite: 2]
| readlines () | Membaca semua baris dan mengembalikannya dalam bentuk list |[cite: 3]
| write (teks) | Menulis teks ke dalam file |[cite: 3]
| writelines (list) | Menulis beberapa baris sekaligus dari list string |[cite: 3]
| close () | Menutup file setelah digunakan |[cite: 3]

### D. Langkah Praktikum
**Latihan 1: Membuat dan Menulis File**[cite: 3]

````python
# membuat dan menulis file
file = open("catatan.txt", "w")
file.write("Halo, ini adalah catatan pertama saya.\n")
file.write("Belajar file handling di Python sangat menarik!\n")
file.close()
print("File berhasil dibuat dan ditulis.")
````

**Latihan 2: Membaca File**[cite: 3]

````python
file = open("catatan.txt", "r")
print("Isi file:")
print(file.read())
file.close()
````

**Latihan 3: Menambahkan Data ke File**[cite: 3]

````python
with open("catatan.txt", "a") as file:
    file.write("Menambahkan baris baru ke file.\n")
print("Data berhasil ditambahkan.")
````

**Latihan 4: Membaca File per Baris**[cite: 3, 4]

````python
with open("catatan.txt", "r") as file:
    for baris in file:
        print(baris.strip())
````

**Latihan 5: Penanganan Kesalahan (Error Handling)**[cite: 4]

````python
try:
    with open("tidak_ada.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("File tidak ditemukan!")
````
